```python
def pull_data(ticker_bbg, start_date, end_date):
    """
    EQY_DVD_YLD_12M is ref-only — no BDH history available via EMDS.
    Pulling current value via ref() and holding constant.
    For Canadian banks this is a reasonable PoC assumption given dividend stability.
    Flag this explicitly if presenting results.
    """
    px_df  = con.bdh(ticker_bbg,    'PX_LAST',                   start_date=start_date, end_date=end_date)
    iv_df  = con.bdh(ticker_bbg,    '30DAY_IMPVOL_100.0%MNY_DF', start_date=start_date, end_date=end_date)
    rf_df  = con.bdh('CORRA Index', 'PX_LAST',                   start_date=start_date, end_date=end_date)

    # Point-in-time div yield — held constant across backtest
    div_ref = con.ref(ticker_bbg, 'EQY_DVD_YLD_12M')
    q_const = float(div_ref['value'].iloc[0]) / 100.0

    data = pd.concat([
        px_df.iloc[:, 0].rename('S'),
        iv_df.iloc[:, 0].rename('sigma'),
        rf_df.iloc[:, 0].rename('r'),
    ], axis=1)

    data['sigma'] = data['sigma'].ffill() / 100.0
    data['r']     = data['r'].ffill()     / 100.0
    data['q']     = q_const

    return data.dropna(subset=['S', 'sigma', 'r'])
```

```python
print(pull_data('BNS CN Equity', '', '2026-01-05'))
```
